## Problem 1: Running Sum of 1D Array
Description: Given an array nums. We define a running sum of an array as: runningSum[i] = sum(nums[0]...nums[i]).
Return the running sum of nums.
Examples: Input: [1,2,3,4] → Output: [1,3,6,10] Input: [1,1,1,1,1]→ Output: [1,2,3,4,5] Input: [3,1,2,10,1]→ Output: [3,4,6,16,17]
Constraints: 1 <= nums.length <= 1000 -10^6 <= nums[i] <= 10^6

```python
from typing import List

class Solution:
    def runningSum(self, nums: List[int]) -> List[int]:
        l=[]
        sum1=0
        for i in nums:
            sum1+=i
            l.append(sum1)
        return l
a=Solution()
result=a.runningSum([1,1,1,1,1])
print(result)
```

### Problem: Shuffle the Array
Given the array nums consisting of 2n elements in the form [x1,x2,...,xn,y1,y2,...,yn].

Return the array in the form [x1,y1,x2,y2,...,xn,yn].

Examples:

Example 1: Input: nums = [2,5,1,3,4,7], n = 3 Output: [2,3,5,4,1,7]

Example 2: Input: nums = [1,2,3,4,4,3,2,1], n = 4 Output: [1,4,2,3,3,2,4,1]

Example 3: Input: nums = [1,1,2,2], n = 2 Output: [1,2,1,2]

Constraints: 1 <= n <= 500 nums.length == 2n 1 <= nums[i] <= 10^3

```python
class Solution:
    def shuffle(self, nums: List[int], n: int) -> List[int]:
        first=nums[:n]
        last=nums[n:]
        lst=[]
        for i in range(n):
            lst.append(first[i])
            lst.append(last[i])
        return lst
a=Solution()
result=a.shuffle([2,5,1,3,4,7],3)
print(result)
```

```python
class Solution:
    def kidsWithCandies(self, candies: List[int], extraCandies: int) -> List[bool]:
        max1=max(candies)
        lst=[]
        for i in candies:
            sum1=i+extraCandies
            if sum1>=max1:
                lst.append(True)
            else:
                lst.append(False)
        return lst
a=Solution()
result=a.kidsWithCandies([2,3,5,1,3],3)
print(result)
```
